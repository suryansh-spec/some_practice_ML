```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
data = pd.read_csv('G:\project\project backup\weather.csv')
data.head(20)
```

```python
ema = data['MaxTemp'].ewm(com=0.3).mean()
```

```python
plt.figure(figsize=(12,8))
plt.plot(data['MaxTemp'], label = "maxtemp")
plt.plot(ema, label = "ema")
plt.xlabel("days")
plt.ylabel("temp")
plt.legend()
plt.show()
```
