```python
x= torch.arange(-5, 5, 0.1).view(-1,1)
w = 0.02
b = 0.3

y = w*x + b

print(x, y)
```

```python
class LinearRegression(nn.Module):
    def __init__(self):
        super().__init__()
        
        #define weights for the regression
        self.weights = nn.Parameter(torch.randn(1,
                                                requires_grad=True, #parameter to be updated by gradient descent enabled
                                                dtype=torch.float))
        #define bias of the regression      
        self.bias = nn.Parameter(torch.randn(1,
                                             requires_grad=True, 
                                             dtype=torch.float))
        
    #forward method to define the computation 
    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.weights * x + self.bias  #main formula for linear regression 
```

```python
torch.manual_seed(42)

model0 = LinearRegression()

list(model0.parameters())
```

```python
model0.state_dict()
```

```python
with torch.inference_mode():
    y_pred = model0(x)

y_pred
```

```python
y
```

```python
lossFn = nn.MSELoss()

optimizer = torch.optim.SGD(model0.parameters(), lr = 0.01)

epochs = 1000
```

```python
for epoch in range(epochs):

    y_pred = model0(x)

    loss = lossFn(y_pred, y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if epoch % 100 == 0:
        print("epoch", {epoch}, "| loss" , loss.item())
        print("weight", model0.weights.item())
        print("bias", model0.bias.item())
```

```python
loss_value = []
epoch_value = []

loss_value.append(loss.item())
epoch_value.append(epoch)
```
